```python
monthly_expense = (
    df.groupby("Month")["Total_Amount"]
    .sum()
)

monthly_expense
```

```python
highest_month = monthly_expense.idxmax()
highest_amount = monthly_expense.max()

lowest_month = monthly_expense.idxmin()
lowest_amount = monthly_expense.min()

print("Highest Expense Month:", highest_month)
print("Highest Expense Amount:", highest_amount)

print("Lowest Expense Month:", lowest_month)
print("Lowest Expense Amount:", lowest_amount)
```

```python
df["Expense_Per_Day"] = df["Total_Amount"] / df["Travel_Days"]

print("Average Expense Per Travel Day:", round(df["Expense_Per_Day"].mean(), 2))
print("Highest Expense Per Day:", round(df["Expense_Per_Day"].max(), 2))
print("Lowest Expense Per Day:", round(df["Expense_Per_Day"].min(), 2))
```

```python
print("Total Advance Payment:", df["Advance_Payment"].sum())
print("Total Travel Expense:", df["Total_Amount"].sum())
print("Total Balance Amount:", df["Balance_Amount"].sum())

print("\nAverage Advance Payment:", round(df["Advance_Payment"].mean(), 2))
print("Average Travel Expense:", round(df["Total_Amount"].mean(), 2))
print("Average Balance Amount:", round(df["Balance_Amount"].mean(), 2))
```

```python
positive_balance = (df["Balance_Amount"] > 0).sum()
negative_balance = (df["Balance_Amount"] < 0).sum()
zero_balance = (df["Balance_Amount"] == 0).sum()

print("Positive Balance Records:", positive_balance)
print("Negative Balance Records:", negative_balance)
print("Zero Balance Records:", zero_balance)
```

```python
client_avg_expense = (
    df.groupby("Client")["Total_Amount"]
      .mean()
      .sort_values(ascending=False)
      .round(2)
)

print(client_avg_expense)
```

```python
top_10_trips = (
    df[["Expense_ID", "Payee_Name", "Client", "Site_City", "Travel_Days", "Total_Amount"]]
    .sort_values("Total_Amount", ascending=False)
    .head(10)
)

print(top_10_trips)
```

```python
expense_categories = [
    "Food_Expense",
    "Lodging_Boarding_Expense",
    "Local_Conveyance_Expense",
    "Petrol_Diesel_Expense",
    "NDT_Charges",
    "Other_Expense"
]

category_percentage = (
    df[expense_categories].sum() / df[expense_categories].sum().sum() * 100
).round(2)

print(category_percentage.sort_values(ascending=False))
```

```python
travel_days_analysis = (
    df.groupby("Travel_Days")
      .agg(
          Total_Expense=("Total_Amount", "sum"),
          Average_Expense=("Total_Amount", "mean"),
          Number_of_Trips=("Expense_ID", "count")
      )
      .sort_index()
      .round(2)
)

print(travel_days_analysis)
```

```python
work_analysis = (
    df.groupby("Work_Done")
      .agg(
          Total_Expense=("Total_Amount", "sum"),
          Average_Expense=("Total_Amount", "mean"),
          Number_of_Trips=("Expense_ID", "count")
      )
      .sort_values("Total_Expense", ascending=False)
      .round(2)
)

print(work_analysis)
```

```python
employee_work_analysis = (
    df.groupby(["Payee_Name", "Work_Done"])["Total_Amount"]
      .agg(["sum", "mean", "count"])
      .round(2)
      .sort_values("sum", ascending=False)
)

print(employee_work_analysis)
```

```python
work_daily_analysis = (
    df.groupby("Work_Done")
      .agg(
          Total_Expense=("Total_Amount", "sum"),
          Total_Travel_Days=("Travel_Days", "sum"),
          Average_Expense_Per_Day=("Total_Amount", "mean")
      )
)

work_daily_analysis["Expense_Per_Travel_Day"] = (
    work_daily_analysis["Total_Expense"] /
    work_daily_analysis["Total_Travel_Days"]
).round(2)

work_daily_analysis = work_daily_analysis.sort_values(
    "Expense_Per_Travel_Day",
    ascending=False
)

print(work_daily_analysis.round(2))
```

```python
employee_efficiency = (
    df.groupby("Payee_Name")
      .agg(
          Total_Expense=("Total_Amount", "sum"),
          Total_Travel_Days=("Travel_Days", "sum"),
          Number_of_Trips=("Expense_ID", "count")
      )
)

employee_efficiency["Expense_Per_Travel_Day"] = (
    employee_efficiency["Total_Expense"] /
    employee_efficiency["Total_Travel_Days"]
).round(2)

employee_efficiency = employee_efficiency.sort_values(
    "Expense_Per_Travel_Day",
    ascending=False
)

print(employee_efficiency.round(2))
```

```python
site_efficiency = (
    df.groupby("Site_City")
      .agg(
          Total_Expense=("Total_Amount", "sum"),
          Total_Travel_Days=("Travel_Days", "sum"),
          Number_of_Trips=("Expense_ID", "count")
      )
)

site_efficiency["Expense_Per_Travel_Day"] = (
    site_efficiency["Total_Expense"] /
    site_efficiency["Total_Travel_Days"]
).round(2)

site_efficiency = site_efficiency.sort_values(
    "Expense_Per_Travel_Day",
    ascending=False
)

print(site_efficiency)
```

```python
site_category_expense = df.groupby("Site_City")[
    [
        "Food_Expense",
        "Lodging_Boarding_Expense",
        "Local_Conveyance_Expense",
        "Petrol_Diesel_Expense",
        "NDT_Charges",
        "Other_Expense"
    ]
].sum()

print(site_category_expense)
```

```python
site_cost_efficiency = (
    df.groupby("Site_City")
      .agg(
          Total_Expense=("Total_Amount", "sum"),
          Total_Travel_Days=("Travel_Days", "sum"),
          Number_of_Trips=("Expense_ID", "count"),
          Total_Lodging=("Lodging_Boarding_Expense", "sum"),
          Total_NDT=("NDT_Charges", "sum")
      )
)

site_cost_efficiency["Expense_Per_Day"] = (
    site_cost_efficiency["Total_Expense"] /
    site_cost_efficiency["Total_Travel_Days"]
).round(2)

site_cost_efficiency["Expense_Per_Trip"] = (
    site_cost_efficiency["Total_Expense"] /
    site_cost_efficiency["Number_of_Trips"]
).round(2)

print(site_cost_efficiency.sort_values(
    "Expense_Per_Day",
    ascending=False
))
```

```python
overall_daily_cost = df["Total_Amount"].sum() / df["Travel_Days"].sum()

print("Overall Average Expense Per Travel Day:", round(overall_daily_cost, 2))

site_cost_efficiency["Extra_Cost_Per_Day"] = (
    site_cost_efficiency["Expense_Per_Day"] - overall_daily_cost
).round(2)

site_cost_efficiency["Potential_Extra_Cost"] = (
    site_cost_efficiency["Extra_Cost_Per_Day"] *
    site_cost_efficiency["Total_Travel_Days"]
).round(2)

print(
    site_cost_efficiency[
        [
            "Expense_Per_Day",
            "Total_Travel_Days",
            "Extra_Cost_Per_Day",
            "Potential_Extra_Cost"
        ]
    ].sort_values(
        "Potential_Extra_Cost",
        ascending=False
    )
)
```

```python
categories = [
    "Food_Expense",
    "Lodging_Boarding_Expense",
    "Local_Conveyance_Expense",
    "Petrol_Diesel_Expense",
    "NDT_Charges",
    "Other_Expense"
]

site_category_daily = (
    df.groupby("Site_City")[categories].sum()
    .div(df.groupby("Site_City")["Travel_Days"].sum(), axis=0)
)

site_category_daily = site_category_daily.round(2)

print(site_category_daily)
```

```python
category_trip_efficiency = (
    df.groupby("Site_City")
      .agg(
          Number_of_Trips=("Expense_ID", "count"),
          Total_NDT=("NDT_Charges", "sum"),
          Total_Lodging=("Lodging_Boarding_Expense", "sum")
      )
)

category_trip_efficiency["NDT_Per_Trip"] = (
    category_trip_efficiency["Total_NDT"] /
    category_trip_efficiency["Number_of_Trips"]
).round(2)

category_trip_efficiency["Lodging_Per_Trip"] = (
    category_trip_efficiency["Total_Lodging"] /
    category_trip_efficiency["Number_of_Trips"]
).round(2)

print(
    category_trip_efficiency.sort_values(
        "NDT_Per_Trip",
        ascending=False
    )
)
```

```python
site_category_percentage = (
    df.groupby("Site_City")[
        [
            "Food_Expense",
            "Lodging_Boarding_Expense",
            "Local_Conveyance_Expense",
            "Petrol_Diesel_Expense",
            "NDT_Charges",
            "Other_Expense"
        ]
    ].sum()
)

site_totals = df.groupby("Site_City")["Total_Amount"].sum()

site_category_percentage = (
    site_category_percentage
    .div(site_totals, axis=0)
    .mul(100)
    .round(2)
)

print(site_category_percentage)
```

```python
trip_analysis = df[
    [
        "Expense_ID",
        "Payee_Name",
        "Site_City",
        "Client",
        "Travel_Days",
        "Food_Expense",
        "Lodging_Boarding_Expense",
        "Local_Conveyance_Expense",
        "Petrol_Diesel_Expense",
        "NDT_Charges",
        "Other_Expense",
        "Total_Amount"
    ]
].copy()

trip_analysis["Expense_Per_Day"] = (
    trip_analysis["Total_Amount"] /
    trip_analysis["Travel_Days"]
).round(2)

print(
    trip_analysis.sort_values(
        "Expense_Per_Day",
        ascending=False
    ).head(10)
)
```

```python
Q1 = df["Total_Amount"].quantile(0.25)
Q3 = df["Total_Amount"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = df[
    (df["Total_Amount"] < lower_limit) |
    (df["Total_Amount"] > upper_limit)
]

print("\nNumber of Outlier Trips:", len(outliers))

print(
    outliers[
        [
            "Expense_ID",
            "Payee_Name",
            "Site_City",
            "Travel_Days",
            "Total_Amount"
        ]
    ].sort_values(
        "Total_Amount",
        ascending=False
    )
)
```

```python
employee_site = (
    df.groupby(["Payee_Name", "Site_City"])["Total_Amount"]
    .agg(
        Total_Expense="sum",
        Number_of_Trips="count",
        Average_Expense="mean"
    )
    .sort_values("Total_Expense", ascending=False)
)

print(employee_site)
```

```python
df["Advance_Utilization_%"] = (
    df["Total_Amount"] / df["Advance_Payment"]
) * 100

print(
    df[
        [
            "Expense_ID",
            "Payee_Name",
            "Advance_Payment",
            "Total_Amount",
            "Balance_Amount",
            "Advance_Utilization_%"
        ]
    ].head(10)
)
```

```python
total_advance = df["Advance_Payment"].sum()
total_expense = df["Total_Amount"].sum()

overall_utilization = (total_expense / total_advance) * 100

print("Total Advance:", total_advance)
print("Total Expense:", total_expense)
print("Overall Advance Utilization:", round(overall_utilization, 2), "%")
```

```python
monthly_advance = (
    df.groupby("Month")
    .agg(
        Total_Advance=("Advance_Payment", "sum"),
        Total_Expense=("Total_Amount", "sum")
    )
)

monthly_advance["Difference"] = (
    monthly_advance["Total_Expense"] -
    monthly_advance["Total_Advance"]
)

monthly_advance["Utilization_%"] = (
    monthly_advance["Total_Expense"] /
    monthly_advance["Total_Advance"] * 100
).round(2)

print(monthly_advance)
```

```python
correlation = df[
    [
        "Travel_Days",
        "Food_Expense",
        "Lodging_Boarding_Expense",
        "Local_Conveyance_Expense",
        "Petrol_Diesel_Expense",
        "NDT_Charges",
        "Other_Expense",
        "Total_Amount"
    ]
].corr()

print(correlation["Total_Amount"].sort_values(ascending=False))
```

```python
expense_bins = [0, 5000, 10000, 15000, 20000, 25000, float("inf")]

expense_labels = [
    "0-5K",
    "5K-10K",
    "10K-15K",
    "15K-20K",
    "20K-25K",
    "25K+"
]

df["Expense_Range"] = pd.cut(
    df["Total_Amount"],
    bins=expense_bins,
    labels=expense_labels,
    right=False
)

expense_distribution = df["Expense_Range"].value_counts().sort_index()

print(expense_distribution)
```

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

monthly_expense = (
    df.groupby("Month", observed=False)["Total_Amount"]
    .sum()
    .reindex(month_order)
)

print(monthly_expense)
```

```python
import matplotlib.pyplot as plt

monthly_expense = monthly_expense.dropna()

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_expense.index,
    monthly_expense.values,
    marker="o"
)

plt.title("Monthly Travel Expense Trend")
plt.xlabel("Month")
plt.ylabel("Total Expense (₹)")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
employee_expense = (
    df.groupby("Payee_Name")["Total_Amount"]
    .sum()
    .sort_values()
)

plt.figure(figsize=(10, 6))

employee_expense.plot(kind="barh")

plt.title("Total Travel Expense by Employee")
plt.xlabel("Total Expense (₹)")
plt.ylabel("Employee")

plt.grid(axis="x")
plt.show()
```

```python
site_expense = (
    df.groupby("Site_City")["Total_Amount"]
    .sum()
    .sort_values()
)

plt.figure(figsize=(10, 6))

site_expense.plot(kind="barh")

plt.title("Total Travel Expense by Site")
plt.xlabel("Total Expense (₹)")
plt.ylabel("Site")

plt.grid(axis="x")
plt.show()
```

```python
site_daily_expense = (
    df.groupby("Site_City")
    .apply(
        lambda x: x["Total_Amount"].sum() / x["Travel_Days"].sum(),
        include_groups=False
    )
    .sort_values()
)

plt.figure(figsize=(10, 6))

site_daily_expense.plot(kind="barh")

plt.title("Average Expense per Travel Day by Site")
plt.xlabel("Expense per Travel Day (₹)")
plt.ylabel("Site")

plt.grid(axis="x")
plt.show()
```
